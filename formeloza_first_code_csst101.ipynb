3. OPERATION


```r
x <- 5
y <- 3

add <- x + y
sub <- x - y
multiply <- x * y
divide <- x / y

print (paste0("Add: ", add))
print (paste0("Subtract: ", sub))
print (paste0("Multiply: ", multiply))
print (paste0 ("Divide: ", divide))
```

8. VECTORS

```r
grades <- c(90, 85, 97, 98, 99)

grades

cat("MEAN: ", mean(grades))
cat("\nMAX: ", max(grades))
cat("\nMIN: ",min(grades))
cat("\nSUM: ", sum(grades))
cat("\nLENGTH: ", length(grades))
```
