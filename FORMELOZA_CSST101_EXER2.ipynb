DATA FRAME

```r
#1
student_data_r <- data.frame(
  Student_ID = c('S001', 'S002', 'S003', 'S004', 'S005'),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65),
  Risk = c("Low Risk", "High Risk", "Low Risk", "Moderate Risk", "High Risk")
)

print(head(student_data_r))
```

Creating Variables

```r
#2
grades_list_r <- c(85, 92, 78, 65, 90)
print(paste("Student's grades:", paste(grades_list_r, collapse=", ")))
```

Basic Calculations

```r
#3
total <- sum(grades_list_r)
count <- length(grades_list_r)

avg <- total / count
cat(sprintf("Average grade: %.2f\n", avg))

threshold <- 60
if (avg >= threshold) {
  cat(sprintf("The student passed with an average of %.2f\n", avg))
} else {
  cat(sprintf("The student did not pass with an average of %.2f\n", avg))
}
```

Comparison Operators

```r
#4
attendance_score_r <- 72
individual_grade_r <- 70

print(paste("Attendance score:", attendance_score_r))
print(paste("Individual grade:", individual_grade_r))

print(paste("Attendance < 75:", attendance_score_r < 75))
print(paste("Grade < 75:", individual_grade_r < 75))
print(paste("Attendance >= 75:", attendance_score_r >= 75))
print(paste("Grade >= 75:", individual_grade_r >= 75))
print(paste("Attendance == 72:", attendance_score_r == 72))
print(paste("Grade != 70:", individual_grade_r != 70))
```

 Using AND and OR

```r
#5
both_risk <- (attendance_score_r < 75) & (individual_grade_r < 75)
print(paste0("Both attendance (", attendance_score_r, ") AND grade (", individual_grade_r, ") are below 75? ", both_risk))

either_risk <- (attendance_score_r < 75) | (individual_grade_r < 75)
print(paste0("Either attendance (", attendance_score_r, ") OR grade (", individual_grade_r, ") is below 75? ", either_risk))
```

If-Else Statement

```r
#6
check_grade <- individual_grade_r
threshold <- 75

print(paste("Student's grade:", check_grade))

if (check_grade >= threshold) {
  print(paste("Grade >=", threshold, ", PASS"))
} else {
  print(paste("Grade <", threshold, ", FAIL"))
}
```

Multiple Conditions (`if-else if-else`)

```r
#7
if (check_grade >= 90 && check_grade <= 100) {
  print("Classification: Excellent")
} else if (check_grade >= 85 && check_grade <= 89) {
  print("Classification: Very Good")
} else if (check_grade >= 80 && check_grade <= 84) {
  print("Classification: Good")
} else if (check_grade >= 75 && check_grade <= 79) {
  print("Classification: Passed")
} else {
  print("Classification: Failed")
}
```

 Creating a Function

```r
#8
classify <- function(grade) {
  if (grade >= 90 && grade <= 100) {
    return("Excellent")
  } else if (grade >= 85 && grade <= 89) {
    return("Very Good")
  } else if (grade >= 80 && grade <= 84) {
    return("Good")
  } else if (grade >= 75 && grade <= 79) {
    return("Passed")
  } else {
    return("Failed")
  }
}

result <- classify(check_grade)
print(paste("Classification:", result))
```

 Creating a Simple Rule-Based System (Risk Category)

```r
#11
s_data$Risk <- ifelse((s_data$Attendance < 75) & (s_data$Grade < 75), 'High Risk',
               ifelse((s_data$Attendance < 80) | (s_data$Grade < 80), 'Moderate Risk',
               ifelse((s_data$Attendance >= 85) & (s_data$Grade >= 85), 'Low Risk',
               'Low Risk')))
print(head(s_data))
```

CREATING A SIMPLE RULE-BASED SYSTEM

```r
#12
check_risk <- function(att, grd) {
    if ((att < 75) && (grd < 75)) {
        return('High Risk')
    } else if ((att < 80) || (grd < 80)) {
        return('Moderate Risk')
    } else if ((att >= 85) && (grd >= 85)) {
        return('Low Risk')
    } else {
        return('Low Risk')
    }
}
```

Another DataFrame and Risk Classification with Function Application

```r
#13
s_data$Risk <- mapply(
    check_risk,
    s_data$Attendance,
    s_data$Grade
)
print(s_data)
```

MINI KRR SYSTEM

```r
#17
krr <- function(id, att, grd) {
  if (att < 75 && grd < 75) {
    risk <- "High Risk"
  } else if (att < 80 || grd < 80) {
    risk <- "Moderate Risk"
  } else {
    risk <- "Low Risk"
  }

  if (risk == "High Risk") {
    action <- "immediate academic intervention"
  } else if (risk == "Moderate Risk") {
    action <- "academic monitoring"
  } else {
    action <- "continue regular monitoring"
  }

  print(paste("Student ID:", id))
  print(paste("Attendance:", att, "Grade:", grd))
  print(paste("What is the academic risk of this student?", risk))
  print(paste("What intervention should be provided?", action))
}

krr("S002", 72, 70)
```

FINAL SYSTEM
