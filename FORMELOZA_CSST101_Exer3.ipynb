7. Implement the Rules

Rule 1: Eligible - Programming Grade >= 80 AND Math Grade >= 75 AND Attendance >= 80% AND Prerequisite = Yes.

Rule 2: Conditionally Eligible - Programming Grade >= 75 AND Math Grade >= 70 AND Attendance >= 75% but Rule 1 is not satisfied.

Rule 3: Not Eligible - if neither Rule 1 or Rule 2 is satisfied, classify the students as Not ELigible.

```r
check_eligibility <- function(programming,math, attendance, prereq) {

  if (programming >= 80 &&
    math >= 75 &&
    attendance >= 80 &&
    prereq == "Yes") {
      return ("Eligible")

    } else if (programming >= 75 &&
      math >= 70 &&
      attendance >= 75) {
        return("Conditional Eligible")

      }else{
        return("Not Eligible")
      }
}
```

8. Check Eligibility

```r
students$Eligibility <- mapply(
  check_eligibility,
  students$Programming_Grade,
  students$Math_Grade,
  students$Attendance,
  students$Prerequisite
)

students
```

9. Create Recommendation Rules

1. Eligible - Student may enroll

2. Conditionally Eligible - Consult Academic Adviser

3. Not Eligible - Complete Requirements before enrollment

```r
generate_recommendation <- function(eligibility) {
  if (eligibility == "Eligible") {
    return("Student may enroll")
  } else if (eligibility == "Conditional Eligible") {
    return("Consult Academic Adviser")
  } else {
    return("Complete Requirements before enrollment")
  }
}
```

Apply the Functions to All Students

```r
students$Recommendation <- mapply(
  generate_recommendation,
  students$Eligibility
)

students
```
